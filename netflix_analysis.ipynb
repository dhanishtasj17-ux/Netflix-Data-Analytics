```python
df_clean = df_clean.copy()

df_clean["date_added"] = pd.to_datetime(
    df_clean["date_added"],
    format="mixed",
    errors="coerce"
)

df_clean["year_added"] = df_clean["date_added"].dt.year
```

```python
df_clean["year_added"].value_counts().sort_index().plot(kind="line", marker="o")
plt.title("Netflix Titles Added Each Year")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")
plt.show()
```

```python
directors = df_clean["director"].dropna().str.split(", ").explode()

directors.value_counts().head(10)
```

```python
directors.value_counts().head(10).plot(kind="barh")
plt.title("Top 10 Directors on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Director")
plt.show()
```

```python
actors = df_clean["cast"].dropna().str.split(", ").explode()

actors.value_counts().head(10)
```

```python
actors.value_counts().head(10).plot(kind="barh")
plt.title("Top 10 Actors on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Actor")
plt.show()
```

```python
movies = df_clean[df_clean["type"] == "Movie"].copy()

movies["duration_minutes"] = movies["duration"].str.extract(r"(\d+)").astype(float)

movies["duration_minutes"].describe()
```

```python
movies["duration_minutes"].plot(kind="hist", bins=20)
plt.title("Distribution of Movie Durations")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Movies")
plt.show()
```

```python
tv_shows = df_clean[df_clean["type"] == "TV Show"].copy()

tv_shows["seasons"] = tv_shows["duration"].str.extract(r"(\d+)").astype(float)

tv_shows["seasons"].value_counts().sort_index()
```

```python
tv_shows["seasons"].value_counts().sort_index().plot(kind="bar")
plt.title("Number of Seasons for Netflix TV Shows")
plt.xlabel("Number of Seasons")
plt.ylabel("Number of TV Shows")
plt.show()
```

```python
titles_by_year_type = (
    df_clean.groupby(["year_added", "type"])
    .size()
    .unstack(fill_value=0)
)

titles_by_year_type.plot(kind="line", marker="o")
plt.title("Netflix Titles Added Each Year by Type")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")  
plt.legend(title="Type")
plt.show()
```

```python
df_clean.to_csv("netflix_titles_cleaned.csv", index=False)
```
